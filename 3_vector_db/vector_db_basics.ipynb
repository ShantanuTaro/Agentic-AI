```python
import chromadb

client = chromadb.Client()
```

```python
collection = client.create_collection("news_article")
```

```python
collection.add(
    ids=["id1","id2", "id3", "id4"],
    documents=[
        "Apple is leading phone company",
        "Tesla booked profit",
        "Apples are high in fiber and vitamin C",
        "SpaceX got NASA contract"
    ]
)
```

```python
data = collection.get(
    include = ["documents", "embeddings", "metadatas"]
)

for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text:", data["documents"][i])
    print("Embedding length:", len(emb))
    print("First 10 values", emb)
```

```python
results = collection.query(
    query_texts = ["Query related to health"],
    n_results = 1
)
print(results)
```
